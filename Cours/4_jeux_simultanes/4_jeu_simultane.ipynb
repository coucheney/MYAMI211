# Jeux à $k$ joueurs simultanés



### Un premier exemple à 2 joueurs: la guerre des sexes

---

|  | **Opéra**  | **Foot**

|**Opéra**  |  (2, 3) | (0, 0)  | 
|**Foot**  |  (1, 1) | (3, 2)  | 



### Modèle général

---

* ensemble fini de joueurs $\{1, \dots, K\}$
* chaque joueur $k$ a un ensemble fini d'actions $\mathcal{S}_k$
* chaque joueur a une fonction de gain (ou utilité) $U_k: \mathcal{S} \rightarrow \mathbb{R}$ qu'ils cherchent à maximiser, avec $\mathcal{S} = \mathcal{S}_1\times \mathcal{S}_2\times \dots \times \mathcal{S}_K$

* On note $\Delta_k$ l'ensemble des stratégies mixtes du joueur $k$.
* Etant donné le **profil de stratégies** $(p_1, p_2, \dots, p_K)$ avec $p_k \in \Delta_k$ pour tout $k \in \{1, \dots, K \}$, on étend la fonction de gain à un tel profil en considérant le gain moyen donné par
$$U_k(p_1, p_2, \dots, p_K) = \sum_{s \in \mathcal{S}}\mathbb{P}[S = s]U_k(s) $$
et 
$$ \mathbb{P}[S = (s_1, s_2, \dots, s_k)] = p_{1,s_1}p_{2,s_2}\dots p_{K,s_K}$$

### Terminologie

---

**Profil de stratégies**: donnée d'une stratégie par joueur. L'ensemble des profils de stratégie est noté $\Delta = \Delta_1\times \Delta_2\times \dots \times \Delta_K$


**Profil de stratégie adverse du joueur k**: ce sont les profils de stratégies qui ne contiennent pas la stratégie du joueur $k$. On note $\Delta_{-k}$ l'ensemble des stratégies adverses du joueur $k$, et $p_{-k}$ une telle stratégie.

Un profil de stratégie peut alors s'écrire $(p_k, p_{-k})$.

### Réponse optimale et équilibre de Nash

---


Etant donné un profil de stratégie adverse $p_{-k}$ pour le joueur $k$, une réponse optimale est une stratégie qui maximise la fonction **linéaire**

$$p_k \rightarrow U_k(p_k, p_{-k})$$



On note $\text{Opt}_k(p)$ l'ensemble des réponses optimales du joueur $k$ face au profil de stratégies adverses $p_{-k}$. Il s'agit d'une fonction multivaluée, ou **correspondance**.



Un **équilibre de Nash** (ou EN) est un profil de stratégies où aucun joueur n'a intérêt à changer de stratégie unilatéralement.

Autrement dit, $p\in \Delta$ est un EN si et seulement si 
 $$U_k(p_k, p_{-k}) \geq U_k(p'_k, p_{-k}), \forall k \in \{1, \dots, K \}, \forall p'_k \in \Delta_k$$
ou
$$p_k \in \text{Opt}_k(p), \forall k \in \{1, \dots, K \}$$
ou vectoriellement
$$p \in \text{Opt}(p) $$
donc si et seulement si $p$ est un **point fixe** de la correspondance $\text{Opt}$.

**Question**: Montrer que dans un jeu à deux joueurs et à somme nulle, les équilibres de Nash et les équilibres minimax coincident.


### Théorème de Nash

---

Dans un jeu sous forme normale fini, il existe toujours un EN.

La preuve repose sur le résultat suivant (admis).

**Théorème de Brouwer**: une fonction continue $f: C \rightarrow C$ où $C$ est un ensemble convexe et compact admet un point fixe.



<div class="alert-info">
Preuve du théorème de Nash:
</div>
   Consider la fonction continue $f: \Delta \rightarrow \Delta$ telle que pour tout joueur $k$ et action $i$ de ce joueur:
   $$f_{k,i}(p) = \frac{p_{k,i} + d_{k,i}(p)^+}{1 + \sum_j d_{k,j}(p)^+} $$
   où
   $$d_{k,i}(p) = U_k(i, p_{-k}) - U_k(p) $$
   et
   $$ x^+ = \max\{0, x\} $$
   
   Ses points fixes coïncident avec les équilibres de Nash du jeu.
   

### Support d'une stratégie

---

Le support d'une stratégie $p \in \Delta_k$ est l'ensemble des actions $i \in \mathcal{S}_k$ qui ont une probabilité non nulle d'être choisie, *i.e.* telles que $p_{i} > 0$. 

**Exemples**:
* $p = (0.3,0,0.1,0,0.4,0.2)$ a pour support les actions $\{1,3,5,6\}$,
* une **stratégie pure** est une stratégie dont le support est réduit à un seul élément.

**Propriété**: une stratégie $p$ du joueur $k$ est une réponse optimale à la stratégie adverse $p_{-k}$ si et seulement si toutes les actions qui sont dans son support sont des réponses optimales à $p_{-k}$.

**Exemple**: si un joueur fait face au vecteur de gains moyens $(4,0,4,2,-1,4)$, alors toute réponse optimale a pour support uniquement des actions de l'ensemble $\{1,3,6 \}$.

**Question**: démontrer cette propriété.

### Jeu à 2 joueurs: tester s'il existe un EN avec une paire de supports donnée
---

Soit $T_1 \subseteq S_1$ et $T_2 \subseteq S_2$ des supports des joueurs 1 et 2 respectivement. Alors toute paire de stratégies  $(p_1,p_2)$ de l'ensemble défini par les **contraintes linéaires** suivantes est un EN dont le support est $T_1 \times T_2$:

* pour tout $s\in T_1$ et tout $s' \in S_1$,
    $$U_1(s, p_2)\geq U_1(s', p_2)$$ 

* pour tout $s\in T_2$ et tout $s' \in S_2$,
    $$U_2(p_1, s) \geq U_2(p_1, s')$$
    
* $p_{1, s} = 0$ si $s \notin T_1$

* $p_{2, s} = 0$ si $s \notin T_2$

* $p_1 \in \Delta_1$

* $p_2 \in \Delta_2$

Noter que cet ensemble peut être vide.

En particulier, $U_1(s, p_2)$ doit être constant pour toute stratégie $s$ dans le support $T_1$, et idem pour le support $T_2$.

### Jeu à 2 joueurs: calcul de l'ensemble des ENs par énumération des supports
---


Pour chaque support de chaque joueur:


* calculer les stratégies des adversaires qui égalisent les gains associés aux actions dans le support
* si elles existent, vérifier si ces gains sont bien optimaux par rapport aux actions en dehors du support
* si c'est le cas, retourner ces stratégies

**Exemple**:

$$
\left(
\begin{array}{c|c}
3,3& 3, 2\\
\hline
2, 2&5, 6\\
\hline
0,3 & 6,1  \\
\end{array}
\right)
$$

Le nombre de supports pour chaque joueur ayant $n$ actions est $2^n-1$.

L'algorithme de Lemke-Howson permet de trouver un EN par méthode de pivotage du type simplexe, mais sa complexité reste exponentielle dans le pire cas.

### Quelques résultats de complexité

---


* calcul d'un EN d'un jeu à 2 joueurs: PPAD-complet, aussi difficile que le problème du calcul d'un point fixe de Brouwer.
* les problèmes de décision suivants sont NP-complets:
    * existe-t'il au moins 2 ENs?
    * existe-t'il un EN qui garantit une valeur $v$ au joueur $i$?
* résoudre un jeu à somme nulle à 3 joueurs est aussi difficile que de résoudre un jeu à 2 joueurs à somme quelconque.
* un jeu à 2 joueurs symétrique admet un EN symétrique.
* calculer un EN dans un jeu à 2 joueurs symétrique est aussi difficile qu'un jeu non symétrique.



## Jeux à 2 joueurs simultanés et à somme nulle: un premier exemple

---

Les joueurs A et B choisissent indépendamment un côté d'une pièce (Pile ou Face) et ils révèlent leur choix en même temps. Si ce choix est identique, alors A donne 2 euros à B, sinon B donne 1 euro à A si A a choisi Pile, et 3 euros sinon. 

**Représentation sous forme normale**: le tableau suivant aggrège les informations du jeu, la première valeur est la gain de A (joueur qui choisit la **ligne**), la deuxième valeur est le gain de B (joueur qui choisit la **colonne**).

|  | **Pile**  | **Face**  
|:--- |:---:|:---:
|**Pile**  |  (-2, 2) | (1, -1)  | 
|**Face**  |  (3, -3) | (-2, 2)  | 



La somme des gains dans chaque case du tableau fait 0: il s'agit d'un jeu à somme nulle, et la deuxième valeur peut être omise.


### Modèle général

---

* Deux joueurs appelés "Ligne" et "Colonne" ayant respectivement $n$ et $m$ actions notées $\{1 \dots n \}$ et $\{1 \dots m\}$.
* Chaque joueur choisit une stratégie en secret, et les deux stratégies sont révélées simultanément.
* On note $A_{ij}$ le gain de Ligne s'il choisit la stratégie pure $i \in \{1 \dots n \}$ et Colonne choisit la stratégie pure $j \in \{1 \dots m\}$. Ligne cherche à **maximiser** $A_{ij}$.
* Le gain de Colonne est alors $-A_{ij}$. Colonne cherche à **minimiser** $A_{ij}$.



**Exemple** ($n=2$ et $m=3$):

$$
A =
\left(
\begin{array}{c|c|c}
1&-1&3\\
\hline
5&0&-2\\
\end{array}
\right)
$$

Quelle est le meilleur gain que Ligne peut garantir? Et pour Colonne?


Si on note
* $ \text{maxmin} = \max_{i \in \{1 \dots n\}} \min_{j \in \{1 \dots m\}}  A_{ij}$
* $ \text{minmax} = \min_{j \in \{1 \dots m\}} \max_{i \in \{1 \dots n\}}  A_{ij}$

on a déjà montré que
$$ \text{maxmin} \leq \text{minmax}$$

### Modèle avec des stratégies mixtes

---

On note $p = (p_1, \dots, p_n) \in \Delta_n$ une stratégie mixte de Ligne et $q = (q_1, \dots, q_m) \in \Delta_m$ une stratégie mixte de Colonne, avec
$$\Delta_k = \{x \in \mathbb{R}^k | x \geq 0 \text{ et }\sum_{i=1}^k x_i = 1\} $$




* $\Delta_k$ est un ensemble **convexe** (sans trou entre toute paire de points de $\Delta_k$)
* $\Delta_k$ est un ensemble **compact** (borné et fermé)

Ces deux propriétés font que toute fonction convexe (resp. concave) définie sur cet ensemble admet une seule valeur minimale (resp. maximale).

**Questions**: montrer que le gain moyen de Ligne est 
* $(pA)_j$ si Ligne choisit $p \in \Delta_n$ et Colonne choisit la stratégie pure $j \in \{1 \dots m\}$ ($=(0,\dots 0 ,1 ,0,\dots 0 )$ où le 1 est sur la j-ème coordonnée)
* $(Aq^\intercal)_i$ si Ligne choisit $i \in \{1 \dots n\}$ et Colonne choisit  $q \in \Delta_m$
* $pAq^\intercal$ si Ligne choisit $p \in \Delta_n$ et Colonne choisit $q \in \Delta_m$


### Théorème du minimax de von Neumann

---
<div class="alert-info">
Pour tout jeu à somme nulle sous forme normale, il existe une valeur $v^*$ et des stratégies $p^* \in \Delta_n$ et $q^* \in \Delta_m$ telles que:

* $p^*A q^{\intercal} \geq v^* , \forall q \in \Delta_m$ ($p^*$ garantit le gain $v^*$ pour Ligne)

* $p A q^{*\intercal} \leq v^* , \forall p \in \Delta_n$ ($q^*$ garantit le gain $-v^*$ pour Colonne)
</div>


Les stratégies $p^*$ et $q^*$ sont appelées **stratégies d'équilibre minimax** et $v^*$ est la **valeur** du jeu

Si on note
* $ \text{supinf} = \sup_{p \in \Delta_n} \inf_{q \in \Delta_m}  pAq^\intercal $
* $ \text{infsup} =  \inf_{q \in \Delta_m} \sup_{p \in \Delta_n} pAq^\intercal $

montrer grâce au théorème que
$$p^* A q^{*\intercal}  = v^* = \text{supinf} = \text{infsup}$$

Montrer également l'unicité de $v^*$. Quid de $p^*$ et de $q^*$?

**Preuve**:
$$
\inf_{q \in \Delta_m} \sup_{p \in \Delta_n} pAq^\intercal
\leq \sup_{p \in \Delta_n} pAq^{*\intercal}
\leq v^*
\leq \inf_{q \in \Delta_m}  p^*Aq^\intercal
\leq \sup_{p \in \Delta_n} \inf_{q \in \Delta_m}  pAq^\intercal
\leq \inf_{q \in \Delta_m} \sup_{p \in \Delta_n} pAq^\intercal
$$

### Lien avec la programmation linéaire

---

Montrer que 
$$p^*A q^{\intercal} \geq v^* , \forall q \in \Delta_m$$
équivaut à
$$(p^*A)_j \geq v^*, \forall j \in \{1 \dots m\} $$

La valeur $v^*$ ainsi que $p^*$ et $q^*$ sont donc une solution au système fini d'inégalités linéaires suivantes avec les variables $p, q, v$:

* $(pA)_j \geq v, \forall j \in \{1 \dots m\} $
* $(Aq^{\intercal})_i \leq v, \forall i \in \{1 \dots n\} $
* $p \in \Delta_n$
* $q \in \Delta_m$



On peut donc trouver une solution, si celle-ci existe, en utilisant la progammation linéaire. 

Le théorème du minimax de von Neumann équivaut à l'existence d'une telle solution  (i.e. l'ensemble des contraintes est non vide).

### Preuve du théorème de von Neumann

---

Soit le programme linéaire suivant:
* $ \max v$
* $(pA)_j \geq v, \forall j \in \{1 \dots m\} $
* $p \in \Delta_n$

**Questions** 
* montrer qu'il admet une solution admissible
* calculer son dual

Le dual est
* $ \min w$
* $(Aq^{\intercal})_i \leq w, \forall i \in \{1 \dots n\} $
* $q \in \Delta_m$

Il admet une solution admissible pour les mêmes raisons que le primal. Les deux problèmes ont donc la même valeur notée $v^*$.

Si $p^*$, $q^*$ et $v^*$ sont des solutions des deux programmes linéaires, on voit qu'elles satisfont les contraintes du théorème de von Neumann.

On peut donc résoudre un jeu en résolvant ce programme linéaire et/ou son dual.
